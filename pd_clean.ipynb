```python
import pandas as pd

df=pd.read_csv('/Users/krishnasundarraobarkule/Downloads/google_new.csv')

display(df.head())
```

```python
df["contact"] = (
    df["contact"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)   # remove non-digits
    .str.lstrip("0")                      # remove leading 0
    .apply(lambda x: "91" + x if len(x) == 10 else x)
)
```

```python
# Clean contact numbers
df["phone"] = (
    df["contact"]
    .astype(str)
    .str.replace(r"\D", "", regex=True)   # remove non-digits
    .str.lstrip("0")                      # remove leading zero
    .apply(lambda x: "91" + x if len(x) == 10 else x)
)

# Drop old contact column
df.drop(columns=["contact"], inplace=True)

# Move phone to first column
df = df[["phone"] + [col for col in df.columns if col != "phone"]]

print(df)
```

```python
df.to_csv('/Users/krishnasundarraobarkule/Downloads/la_so_cleaned.csv', index=False)
```
